3. Marcação dos Pontos</span></center>
<hr style="border:1px solid #0077b9;">

script utiliza a biblioteca Shapely para criar e manipular pontos geográficos no espaço, combinada com o GeoPandas para gerenciar essas localizações com seus respectivos sistemas de referência espacial (CRS). Além disso, emprega o NumPy para gerar coordenadas aleatórias uniformes dentro dos limites dos polígonos e tiles analisados

```python
import geopandas as gpd
import numpy as np
import pandas as pd
from shapely.geometry import Point

tile_id = "C56L49"
tile = gpd.read_file(
    "../data/deter_tiles_AMZ.shp"
).query(f"id == '{tile_id}'")

deter = gpd.read_file(
    "../data/Deter_alerts_tiled.gpkg"
)

deter["DATE"] = pd.to_datetime(deter["DATE"])

col_classe = "CLASS" if "CLASS" in deter.columns else "CLASS"

deter_filtrado = deter[
    (deter["DATE"] >= "2023-01-01") & 
    (deter["DATE"] <= "2024-12-31") & 
    (deter[col_classe] == "DESMATAMENTO_CR")
]

df_recorte = gpd.clip(deter_filtrado, tile)

print(f"Polígonos recortados (2023-2024): {len(df_recorte)}")

def gerar_pontos(geom_unida, df_referencia, n, target, nome_classe):
    minx, miny, maxx, maxy = df_referencia.total_bounds
    pts = []
    while len(pts) < n:
        p = Point(np.random.uniform(minx, maxx), np.random.uniform(miny, maxy))
        if geom_unida.contains(p):
            pts.append({"geometry": p, "target": target, "CLASS": nome_classe})
    return gpd.GeoDataFrame(pts, crs=tile.crs)

p_dist = gerar_pontos(
    df_recorte.geometry.unary_union, df_recorte, 100, 1, "Disturbio"
)

p_flor = gerar_pontos(
    tile.geometry.unary_union.difference(df_recorte.geometry.unary_union),
    tile,
    100,
    0,
    "Floresta_Nativa",
)

df_treino_ml = pd.concat(
    [
        p_dist[["geometry", "target", "CLASS"]],
        p_flor[["geometry", "target", "CLASS"]],
    ],
    ignore_index=True,
)

# Adiciona o ID do tile de forma categórica (sem coordenadas numéricas exatas)
df_treino_ml["tile_id"] = tile_id

print("Distribuição das classes:")
print(df_treino_ml["CLASS"].value_counts())
print(df_treino_ml["target"].value_counts())
```

<center><span style="color:#336699; font-size: 100%;">
4. Plotagem do Gráfico</span></center>
<hr style="border:1px solid #0077b9;">

O script utiliza a biblioteca Folium para criar um mapa interativo com imagem de satélite de fundo, adicionando o contorno azul do tile, os polígonos rosas de desmatamento e os pontos geográficos coloridos (bege para distúrbio e laranja para floresta nativa).

```python
import folium
import pandas as pd

df_recorte_mapa = df_recorte.copy()

for col in df_recorte_mapa.columns:
    if pd.api.types.is_datetime64_any_dtype(df_recorte_mapa[col]):
        df_recorte_mapa[col] = df_recorte_mapa[col].dt.strftime("%Y-%m-%d")

centro = [
    tile.geometry.centroid.y.iloc[0],
    tile.geometry.centroid.x.iloc[0],
]

m = folium.Map(location=centro, zoom_start=11)

folium.TileLayer(
    tiles="https://server.arcgisonline.com/ArcGIS/rest/services/World_Imagery/MapServer/tile/{z}/{y}/{x}",
    attr="Esri",
    name="Satélite",
).add_to(m)

folium.GeoJson(
    tile,
    name="Tile C56L49",
    style_function=lambda x: {
        "fillColor": "transparent",
        "color": "#FFFF00",
        "weight": 3,
        "fillOpacity": 0.1,
    },
    tooltip="Tile C56L49"
).add_to(m)

folium.GeoJson(
    df_recorte_mapa,
    name="Desmatamento DETER (2023-2024)",
    style_function=lambda x: {
        "fillColor": "#ff69b4",
        "color": "black",
        "weight": 0.5,
        "fillOpacity": 0.5,
    },
    tooltip=folium.GeoJsonTooltip(
        fields=["DATE", "CLASS"] if "DATE" in df_recorte_mapa.columns else ["CLASS"],
        aliases=["Data do Alerta:", "Classe:"] if "DATE" in df_recorte_mapa.columns else ["Classe:"]
    )
).add_to(m)

area_interna = tile.geometry.iloc[0].buffer(-0.005)

p_dist_filtrado = p_dist[p_dist.geometry.within(area_interna)]
p_flor_filtrado = p_flor[p_flor.geometry.within(area_interna)]

grupo_disturbio = folium.FeatureGroup(name="Pontos: Distúrbio (100 pts)")
for idx, row in p_dist_filtrado.iterrows():
    folium.CircleMarker(
        location=[row.geometry.y, row.geometry.x],
        radius=2.0,
        color="#EED9C4",
        weight=0.8,
        fill=True,
        fill_color="#EED9C4",
        fill_opacity=1.0,
        popup="Classe: 1 (Distúrbio) | Período: 2023-2024",
    ).add_to(grupo_disturbio)
grupo_disturbio.add_to(m)

grupo_floresta = folium.FeatureGroup(name="Pontos: Floresta Nativa (100 pts)")
for idx, row in p_flor_filtrado.iterrows():
    folium.CircleMarker(
        location=[row.geometry.y, row.geometry.x],
        radius=2.0,
        color="#FF3300",
        weight=0.8,
        fill=True,
        fill_color="#FF6600",
        fill_opacity=1.0,
        popup="Classe: 0 (Floresta Nativa)",
    ).add_to(grupo_floresta)
grupo_floresta.add_to(m)

folium.LayerControl(collapsed=False).add_to(m)

m
```
